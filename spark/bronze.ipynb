```python
from pyspark.sql import functions as F

spark.sql("CREATE SCHEMA IF NOT EXISTS shoplive.bronze")

customers_df = spark.read.option("header", "true").option("inferSchema", "true").csv("/Volumes/shoplive/core/raw/customers.csv")
orders_df = spark.read.option("header", "true").option("inferSchema", "true").csv("/Volumes/shoplive/core/raw/orders_2026_h1.csv")
products_df = spark.read.option("header", "true").option("inferSchema", "true").csv("/Volumes/shoplive/core/raw/products.csv")
events_df = spark.read.option("multiline", "true").json("/Volumes/shoplive/core/event/*.json")

print(f"Customers: {customers_df.count():,} | Orders: {orders_df.count():,} | Products: {products_df.count():,} | Events: {events_df.count():,}")
```

```python
customers_df.write.format("delta").mode("overwrite").saveAsTable("shoplive.bronze.customers")
orders_df.write.format("delta").mode("overwrite").saveAsTable("shoplive.bronze.orders")
products_df.write.format("delta").mode("overwrite").saveAsTable("shoplive.bronze.products")
events_df.write.format("delta").mode("overwrite").saveAsTable("shoplive.bronze.events")

print("All bronze tables created successfully")
```
