```python
from pyspark.sql import SparkSession

spark = SparkSession.builder.appName("bronze layer").getOrCreate()
```

```python
customers_df = spark.read \
    .option("header", "true") \
    .option("inferSchema", "true") \
    .csv("/Volumes/shoplive/core/raw/customers.csv")

orders_df = spark.read \
    .option("header", "true") \
    .option("inferSchema", "true") \
    .csv("/Volumes/shoplive/core/raw/orders_2026_h1.csv")

products_df = spark.read \
    .option("header", "true") \
    .option("inferSchema", "true") \
    .csv("/Volumes/shoplive/core/raw/products.csv")
```

```python
display(customers_df)
```

```python
display(orders_df)
```

```python
display(products_df)
```

```python
customers_df.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("shoplive.bronze.customers")
```

```python
orders_df.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("shoplive.bronze.orders")
```

```python
products_df.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("shoplive.bronze.products")
```
